```python
import pandas as pd
```

```python
_raw = "raw_cc_cs.txt"
_geo = "geo_corr.txt"
_tref = "tref_info.txt"
_iso = "iso_corr_puzzle.txt"
_iso_raw = "iso_corr_raw.txt"


df_raw = pd.read_csv(_raw, sep="\t",names=["exp","target","run","energy","incoming_ions","central","borders"])
df_geo = pd.read_csv(_geo, sep="\t",names=["exp","target","run","energy","xmin","xmax","ymin","ymax"])
df_tref = pd.read_csv(_tref, sep="\t",names=["exp","target","run","energy","tref_number"])
df_iso = pd.read_csv(_iso,sep="\t",names=["exp","target","run","energy","ratio_11c","some1","some2"])
df_iso_raw = pd.read_csv(_iso_raw,sep="\t",names=["exp","target","run","energy","ratio_11c_raw","some1_raw","some2_raw"])
```

```python
print(df_tref)
```

```python
print(df_iso)
```

```python


dfs = [df_raw,df_geo,df_tref,df_iso,df_iso_raw]
from functools import reduce

merge_keys = ["exp", "target", "run", "energy"]



df_merged = reduce(
    lambda left, right: left.merge(right, on=merge_keys, how="outer"),
    dfs
)
```

```python
print(df_merged)
```

```python
df_merged["pc_raw_cc"] = (1-df_merged["central"]/df_merged["incoming_ions"])*100
df_merged["pc_borders"] = (df_merged["borders"]/df_merged["incoming_ions"])*100
df_merged["pc_tref"] = (df_merged["tref_number"]/df_merged["incoming_ions"])*100
#df_merged["pc_geo"] = (df_merged["central"]/df_merged["incoming_ions"])*(0.5*(df_merged["xmin"]+df_merged["xmax"])*0.5*(df_merged["ymin"]+df_merged["ymax"]))*100
df_merged["pc_geo"] = 0.5*(df_merged["xmin"]+df_merged["xmax"])*0.5*(df_merged["ymin"]+df_merged["ymax"])*100
df_merged["pc_iso"] = df_merged["ratio_11c"]*(100 - df_merged["pc_raw_cc"])
df_merged["pc_iso_raw"] = df_merged["ratio_11c_raw"]*(100 - df_merged["pc_raw_cc"])
```

```python
print(df_merged["pc_raw_cc"])
df_merged["pc_geo"] = df_merged["pc_raw_cc"] - df_merged["pc_geo"]* df_merged["pc_raw_cc"]/100
print(df_merged["pc_geo"])
```

```python
print(df_merged["pc_iso"])
```

```python
print(df_merged["pc_iso_raw"])
```

```python
df_final = df_merged[["target", "run", "energy", "pc_raw_cc", "pc_borders","pc_tref","pc_geo","pc_iso","pc_iso_raw"]]
```

```python
print(df_final)
```

```python


category_order = ["c_54","c_1086","c_2198","empty"]
df2 = df_final.copy()

# assign category with .loc to avoid the warning
df2.loc[:, "target"] = pd.Categorical(
    df2["target"],
    categories=category_order,
    ordered=True
)
print(df2)
# now sort without warnings
df_sorted = df2.sort_values(by=["energy", "target"])


print(df_sorted)

print(df_final["target"].unique().tolist())
print(df_final.columns.tolist())
```
